```python
product_info = pd.DataFrame({
"Product Name": [
"Laptop",
"Mouse",
"Printer",
"Tablet",
"Monitor"
],
"Supplier": [
"Dell",
"Logitech",
"HP",
"Samsung",
"LG"
],
"Discount": [
10,
5,
15,
8,
12
]
})
```

```python
print("PRODUCT INFORMATION")
display(product_info)
merged_df = pd.merge(
df,
product_info,
on="Product Name",
how="left"
)
```

```python
print("MERGED DATA")
display(merged_df)
```

```python
region_info = pd.DataFrame(
{
"Region_Manager": [
"Manager A",
"Manager B",
"Manager C",
"Manager D"
]
},
index=[
"North",
"South",
"East",
"West"
]
)
```

```python
print("REGION INFORMATION")
display(region_info)
```

```python
df_region = df.set_index("Region")
joined_df = df_region.join(region_info)
print("JOINED DATA")
display(joined_df)
```

```python
region_sales = df.groupby(
"Region"
)["Sales"].sum()
print("TOTAL SALES BY REGION")
display(region_sales)
```

```python
pivot_table = pd.pivot_table(
df,
values="Sales",
index="Category",
columns="Region",
aggfunc="sum",
fill_value=0
)
print("PIVOT TABLE")
display(pivot_table)
```
