```python
import pandas as pd

data = {
    "StudyHours": [1, 2, 3, 4],
    "PreviousMarks": [2, 3, 5, 4]
}

df = pd.DataFrame(data)
print("original Dataset:")

print(df)
```

```python
scaler = StandardScaler()

# Standardize the data
standardized_data = scaler.fit_transform(df)

# Create a DataFrame
standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

# Print only standardized data
print("Standardized Dataset:")
print(standardized_df.round(6))

```

```python

covariance_matrix = np.cov(
    standardized_data,
    rowvar=False
)

print("\nCovariance Matrix:")
print(covariance_matrix.round(6))
```

```python


eigenvalues, eigenvectors = np.linalg.eig(covariance_matrix)
eigenvalues = eigenvalues.real
eigenvectors = eigenvectors.real

print("Eigenvalues:")
print(eigenvalues.round(6))

print("\nEigenvectors:")
print(eigenvectors.round(6))
```

```python

# Sort eigenvalues in ascending order
index = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[index]
eigenvectors = eigenvectors[:, index]

print("Eigenvalues in Ascending Order:")
print(eigenvalues.round(6))

print("\nEigenvectors in Ascending Order:")
print(eigenvectors.round(6))
```

```python

# Calculate explained variance ratio
explained_variance_ratio = (
    eigenvalues / np.sum(eigenvalues)
)

print("Explained Variance Ratio:")
print(explained_variance_ratio.round(6))

print("\nExplained Variance Ratio (%):")
print((explained_variance_ratio * 100).round(2))
```

```python

# Calculate cumulative explained variance
cumulative_explained_variance = np.cumsum(
    explained_variance_ratio
)

print("Cumulative Explained Variance:")
print(cumulative_explained_variance.round(6))


```
